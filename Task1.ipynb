```python
#import libraries
import pandas as pd
import numpy as np
```

```python
#load dataset
df = pd.read_csv(r"C:\Users\Admin\OneDrive\Documents\Internship project\Task1\Titanic-Dataset.csv");
```

```python
#show top 5 row for checking data
df.head()
```

```python
#check the datatype of the each column
df.info()
```

```python
#exract the the meaningful values from the dataset
df.describe()
```

```python
#check for the null values in the dataset
df.isnull().sum()
```

```python
#handle missing values of Age column
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
#handle missing values of the cabin column
df['Cabin'] = df['Cabin'].fillna("Unknown")
```

```python
#handle missing values of Fare column
df['Fare'] = df['Fare'].fillna(df['Fare'].median())
```

```python
#check for null values for verification
df.isnull().sum()
```

```python
#Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
#Convert Data Types
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
```

```python
#Rename Columns for Clarity
df.rename(columns={
    'Pclass': 'Passenger_Class',
    'SibSp': 'Siblings_Spouses',
    'Parch': 'Parents_Children'
}, inplace=True)
```

```python
#Validation Report
print("\n=== REPORT ===")
print(f"Original shape: {pd.read_csv(r"C:\Users\Admin\OneDrive\Documents\Internship project\Task1\Titanic-Dataset.csv").shape}")
print(f"Cleaned shape: {df.shape}")
print(f"Missing values remaining: {df.isnull().sum().sum()}")
```

```python
#Save Cleaned Dataset
df.to_csv('titanic_cleaned.csv', index=False)
```
